## 3. Por qué escalar

Antes de correr K-Means, discutan **sin prompt-ear todavía**:

> `Age` va de ~18 a ~70. `Annual Income (k$)` va de ~15 a ~140. `Spending
> Score` va de 1 a 100. K-Means agrupa puntos por **distancia euclidiana**. Si
> corremos el algoritmo con estas tres columnas tal cual, sin ajustar nada,
> ¿qué variable va a dominar la distancia? ¿Por qué?

Ahora pídanle a la IA que estandarice las variables numéricas (`Age`,
`Annual Income (k$)`, `Spending Score (1-100)`) con `StandardScaler` de
`sklearn`, y que les muestre cómo quedan las escalas antes y después.

**Prompt sugerido:**

> Usa `StandardScaler` de scikit-learn para estandarizar las columnas `Age`,
> `Annual Income (k$)` y `Spending Score (1-100)` del DataFrame. Muéstrame la
> media y desviación estándar de cada columna antes y después de escalar.

````python
from sklearn.preprocessing import StandardScaler

columnas_numericas = [
    "Age",
    "Annual Income (k$)",
    "Spending Score (1-100)",
]

print("Medias antes de escalar:")
print(df[columnas_numericas].mean())
print("\nDesviaciones estándar antes de escalar:")
print(df[columnas_numericas].std())

scaler = StandardScaler()
df_escalado = pd.DataFrame(
    scaler.fit_transform(df[columnas_numericas]),
    columns=columnas_numericas,
    index=df.index,
)

print("\nMedias después de escalar:")
print(df_escalado.mean())
print("\nDesviaciones estándar después de escalar:")
print(df_escalado.std())
````

**Para discutir:** ¿su predicción de qué variable iba a dominar se confirmó?

## 4. Ver al algoritmo aprender, paso a paso (k=2)

Antes de correr K-Means "de verdad" sobre el dataset completo, vamos a verlo
**aprender en cámara lenta**, con solo dos variables (`Annual Income (k$)` y
`Spending Score (1-100)`, ya estandarizadas) y k=2, para que quepa todo en una
gráfica.

**Prompt sugerido — pídanselo en un solo prompt, pero léanlo con cuidado antes
de mandarlo, porque tiene varios pasos:**

> Usando solo las columnas `Annual Income (k$)` y `Spending Score (1-100)` ya
> estandarizadas, y con k=2:
> 1. Elige 2 centroides iniciales al azar (usa una semilla fija) y grafícalos
>    sobre los puntos.
> 2. Haz una sola iteración de K-Means "a mano": asigna cada punto al
>    centroide más cercano (coloréalos), recalcula los centroides como el
>    promedio de su grupo, y grafica el resultado.
> 3. Repite ese paso 4 veces en total, guardando una gráfica por cada
>    iteración, para que yo pueda ver cómo se van moviendo los centroides.
> 4. Dime en qué iteración los centroides prácticamente dejaron de moverse.

````python
import matplotlib.pyplot as plt
import numpy as np

columnas_kmeans = ["Annual Income (k$)", "Spending Score (1-100)"]
puntos = df_escalado[columnas_kmeans].to_numpy()

rng = np.random.default_rng(42)
centroides = puntos[rng.choice(len(puntos), size=2, replace=False)].copy()

fig, ax = plt.subplots(figsize=(7, 5))
ax.scatter(puntos[:, 0], puntos[:, 1], alpha=0.6)
ax.scatter(centroides[:, 0], centroides[:, 1], c=["red", "black"], s=180, marker="X")
ax.set_title("Centroides iniciales")
ax.set_xlabel(columnas_kmeans[0])
ax.set_ylabel(columnas_kmeans[1])
plt.show()

movimientos = []
for iteracion in range(1, 5):
    distancias = ((puntos[:, np.newaxis, :] - centroides[np.newaxis, :, :]) ** 2).sum(
        axis=2
    )
    etiquetas = distancias.argmin(axis=1)
    nuevos_centroides = np.array(
        [puntos[etiquetas == cluster].mean(axis=0) for cluster in range(2)]
    )
    movimiento = np.linalg.norm(nuevos_centroides - centroides, axis=1)
    movimientos.append(movimiento)

    fig, ax = plt.subplots(figsize=(7, 5))
    ax.scatter(
        puntos[:, 0],
        puntos[:, 1],
        c=etiquetas,
        cmap="viridis",
        alpha=0.6,
    )
    ax.scatter(
        nuevos_centroides[:, 0],
        nuevos_centroides[:, 1],
        c=["red", "black"],
        s=180,
        marker="X",
    )
    ax.set_title(f"Iteración {iteracion}")
    ax.set_xlabel(columnas_kmeans[0])
    ax.set_ylabel(columnas_kmeans[1])
    plt.show()

    centroides = nuevos_centroides
    print(f"Iteración {iteracion}: movimiento de los centroides = {movimiento}")

movimientos = np.array(movimientos)
iteracion_estable = next(
    (
        iteracion
        for iteracion in range(1, len(movimientos) + 1)
        if np.all(movimientos[iteracion - 1] < 1e-3)
    ),
    None,
)

if iteracion_estable is None:
    print("Los centroides aún no se estabilizaron tras 4 iteraciones.")
else:
    print(
        f"Los centroides prácticamente dejaron de moverse en la iteración {iteracion_estable}."
    )
````

## 5. Un K-Means completo, con su propio K

Ahora sí, sobre las **tres variables estandarizadas** (`Age`,
`Annual Income (k$)`, `Spending Score (1-100)`), usando `KMeans` de
`sklearn`. Usen el número de clusters que predijeron en la sección 2.

**Prompt sugerido:**

> Corre `KMeans` de scikit-learn con k=[el número que predijeron] sobre las
> tres columnas estandarizadas, con `random_state=42`. Agrega la etiqueta de
> cluster resultante como una nueva columna al DataFrame original (con las
> variables sin escalar). Haz un scatter de `Annual Income (k$)` contra
> `Spending Score (1-100)` coloreado por cluster.

````python
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans

kmeans = KMeans(n_clusters=5, random_state=42, n_init="auto")
df_escalado["Cluster"] = kmeans.fit_predict(df_escalado[columnas_numericas])
df["Cluster"] = df_escalado["Cluster"]

plt.figure(figsize=(8, 6))
plt.scatter(
    df["Annual Income (k$)"],
    df["Spending Score (1-100)"],
    c=df["Cluster"],
    cmap="viridis",
    s=70,
    alpha=0.8,
)
plt.xlabel("Annual Income (k$)")
plt.ylabel("Spending Score (1-100)")
plt.title("Segmentación de clientes con K-Means (k=5)")
plt.colorbar(label="Cluster")
plt.show()

plt.figure(figsize=(8, 6))
plt.scatter(
    df["Age"],
    df["Spending Score (1-100)"],
    c=df["Cluster"],
    cmap="viridis",
    s=70,
    alpha=0.8,
)
plt.xlabel("Age")
plt.ylabel("Spending Score (1-100)")
plt.title("Edad y spending score por cluster")
plt.colorbar(label="Cluster")
plt.show()

df.head()
````

## 6. ¿Por qué ese K y no otro?

Eligieron su K "a ojo" en la sección 2. Ahora dejen que los datos opinen.
Nota: Silueta score promedia las distancias entre clusters y la distancia entre el mismo cluster

**Prompt sugerido:**

> Corre `KMeans` para valores de k de 2 a 10 sobre las tres columnas
> estandarizadas. Para cada k, calcula la inercia (para el método del codo) y
> el `silhouette_score`. Grafica ambas métricas contra k, en dos subplots.

````python
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score

puntos_escalados = df_escalado[columnas_numericas]
valores_k = range(2, 11)
inercia = []
silueta = []

for k in valores_k:
    modelo = KMeans(n_clusters=k, random_state=42, n_init="auto")
    etiquetas_k = modelo.fit_predict(puntos_escalados)
    inercia.append(modelo.inertia_)
    silueta.append(silhouette_score(puntos_escalados, etiquetas_k))

resultados_k = pd.DataFrame(
    {
        "k": list(valores_k),
        "inercia": inercia,
        "silhouette_score": silueta,
    }
)
display(resultados_k)

fig, axes = plt.subplots(1, 2, figsize=(14, 5))

axes[0].plot(resultados_k["k"], resultados_k["inercia"], marker="o")
axes[0].set_title("Método del codo")
axes[0].set_xlabel("Número de clusters (k)")
axes[0].set_ylabel("Inercia")
axes[0].set_xticks(list(valores_k))

axes[1].plot(resultados_k["k"], resultados_k["silhouette_score"], marker="o")
axes[1].set_title("Silhouette score")
axes[1].set_xlabel("Número de clusters (k)")
axes[1].set_ylabel("Silhouette score")
axes[1].set_xticks(list(valores_k))

plt.tight_layout()
plt.show()

mejor_k = resultados_k.loc[resultados_k["silhouette_score"].idxmax(), "k"]
print(f"El mejor silhouette score se obtiene con k={mejor_k}.")
````

2? Si no coinciden, ¿con cuál se quedarían y por qué?
